# Step 2: Calculate Remaining Useful Life (RUL)

Remaining Useful Life (RUL) is the target variable for predictive maintenance.
For each engine, RUL is calculated as the difference between its maximum operating cycle and the current cycle.

```python

# Step 2: Calculate Remaining Useful Life (RUL)

# Find the maximum cycle for each engine
max_cycle = train.groupby("unit")["cycle"].max().reset_index()

# Rename the column
max_cycle.columns = ["unit", "max_cycle"]

# Merge with original dataset
train = train.merge(max_cycle, on="unit", how="left")

# Calculate RUL
train["RUL"] = train["max_cycle"] - train["cycle"]

# Remove temporary column
train.drop(columns=["max_cycle"], inplace=True)

print("RUL column created successfully.")
```

# Step 3: Separate Features & Target

In this step, we separate the predictor variables (features) from the target variable (RUL). The `unit` identifier is retained for grouped feature engineering in later steps, while `RUL` is used as the prediction target.

```python

# Step 3: Separate Features & Target
# Target variable
y = train["RUL"]

# Feature matrix
X = train.drop(columns=["RUL"])

print("Feature Matrix Shape :", X.shape)
print("Target Vector Shape  :", y.shape)
```

# Step 4: Remove Low Variance Features

Low-variance features contain very little information because their values remain nearly constant across observations. These features are removed to reduce dimensionality, improve computational efficiency, and potentially enhance model performance.

```python
from sklearn.feature_selection import VarianceThreshold
```

```python
variances = X.drop(columns=["unit", "cycle"]).var().sort_values()

print(variances)
```

```python
# Sensor and setting features only
feature_cols = X.drop(columns=["unit", "cycle"]).columns
```

```python
# Threshold can be adjusted
threshold = 0.01

selector = VarianceThreshold(threshold=threshold)

selector.fit(X[feature_cols])
```

```python
selected_features = feature_cols[selector.get_support()]
removed_features = feature_cols[~selector.get_support()]

print("Number of Selected Features :", len(selected_features))
print("Number of Removed Features  :", len(removed_features))

print("\nRemoved Features:")
print(list(removed_features))
```

```python
# Keep unit and cycle for future feature engineering
X = pd.concat(
    [
        X[["unit", "cycle"]],
        X[selected_features]
    ],
    axis=1
)

print("Updated Feature Matrix Shape:", X.shape)
```

## Step 5: Cycle-Based Feature Engineering

```python
# Maximum cycle for each engine
max_cycle = X.groupby("unit")["cycle"].transform("max")

# Percentage of life completed
X["cycle_pct"] = X["cycle"] / max_cycle

# Remaining life percentage
X["life_remaining_pct"] = 1 - X["cycle_pct"]
```

```python
X["max_cycle"] = X.groupby("unit")["cycle"].transform("max")
```

```python
X["cycle_pct"] = X["cycle"] / X["max_cycle"]
```

```python
X["life_remaining_pct"] = 1 - X["cycle_pct"]
```

```python
X["engine_age"] = X["cycle"]
```

```python
X[[
    "unit",
    "cycle",
    "max_cycle",
    "cycle_pct",
    "life_remaining_pct",
    "engine_age"
]].head(10)
```

```python
X.drop(columns=["max_cycle"], inplace=True)
```

## Step 6: Rolling Mean Features

```python
sensor_cols = [
    col for col in X.columns
    if col.startswith("s") and col[1:].isdigit()
]

for col in sensor_cols:
    X[f"{col}_roll_mean"] = (
        X.groupby("unit")[col]
        .transform(lambda x: x.rolling(window=5, min_periods=1).mean())
    )
```

```python
len([col for col in X.columns if "roll_mean" in col])
```

```python
roll_mean_cols = [col for col in X.columns if "roll_mean" in col]

X[roll_mean_cols].head()
```

```python
for col in sensor_cols:
    X[f"{col}_roll_std"] = (
        X.groupby("unit")[col]
        .transform(
            lambda x: x.rolling(window=5, min_periods=1).std()
        )
        .fillna(0)
    )
```

## step 7: Rolling Standard Deviation

## Step 8: Lag Features

```python
for col in sensor_cols:
    X[f"{col}_lag1"] = (
        X.groupby("unit")[col]
        .shift(1)
        .fillna(0)
    )
```

```python
lag_cols = [col for col in X.columns if "_lag1" in col]

print(lag_cols)
```

```python
X[lag_cols].head()
```

## Step 8: Delta Features
Delta = Current Value - Previous Value

```python
for col in sensor_cols:
    X[f"{col}_delta"] = X[col] - X[f"{col}_lag1"]
```

```python
delta_cols = [col for col in X.columns if "_delta" in col]

print(delta_cols)
print(len(delta_cols))
```

```python
X[delta_cols].head()
```

```python
from sklearn.preprocessing import StandardScaler

# Save unit separately
unit = X["unit"]

# Remove unit before scaling
X_no_unit = X.drop(columns=["unit"])

# Scale remaining features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_no_unit)

# Convert back to DataFrame
X_scaled = pd.DataFrame(X_scaled, columns=X_no_unit.columns)

# Add unit back
X_scaled.insert(0, "unit", unit.values)

# Add target
X_scaled["RUL"] = y.values
```
